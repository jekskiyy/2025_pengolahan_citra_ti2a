```python
# Upload file dari lokal
from google.colab import files
uploaded = files.upload()

# Import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, LSTM, SimpleRNN
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

# Membaca file CSV
filename = list(uploaded.keys())[0]
data = pd.read_csv(filename)

# Deteksi kolom harga
harga_cols = ['Close', 'Adj Close', 'Price', 'Last', 'Value']
for col in harga_cols:
    if col in data.columns:
        data['Price'] = data[col]
        print(f"Menggunakan kolom harga: {col}")
        break
else:
    raise ValueError("Tidak ditemukan kolom harga yang valid!")

# Konversi tanggal jika ada
if 'Date' in data.columns:
    data['Date'] = pd.to_datetime(data['Date'], errors='coerce')
    data = data.dropna(subset=['Date'])
    data = data.sort_values('Date')

# =========================
# EDA (Exploratory Data Analysis)
# =========================
print("\nStatistik Deskriptif:")
print(data.describe())

plt.figure(figsize=(8, 6))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Korelasi antar Fitur")
plt.show()

plt.figure(figsize=(12, 5))
plt.plot(data['Price'], label='Harga Emas')
plt.title('Harga Emas Seiring Waktu')
plt.xlabel('Index')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()

if 'Volume' in data.columns:
    plt.figure(figsize=(6, 4))
    sns.histplot(data['Volume'], kde=True)
    plt.title('Distribusi Volume Transaksi')
    plt.show()

# =========================
# Persiapan Data untuk Prediksi Regresi
# =========================
features = ['Open', 'High', 'Low', 'Volume']
target = 'Price'

scaler_X = MinMaxScaler()
X_scaled = scaler_X.fit_transform(data[features])

scaler_y = MinMaxScaler()
y_scaled = scaler_y.fit_transform(data[[target]]).flatten()

def create_sequences(X, y, window_size=10):
    Xs, ys = [], []
    for i in range(len(X) - window_size):
        Xs.append(X[i:i+window_size])
        ys.append(y[i+window_size])
    return np.array(Xs), np.array(ys)

window_size = 10
X_seq, y_seq = create_sequences(X_scaled, y_scaled, window_size)

# =========================
# Definisi Model
# =========================
def build_rnn(input_shape):
    model = Sequential([
        SimpleRNN(64, input_shape=input_shape),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1, activation='linear')
    ])
    model.compile(loss='mean_squared_error', optimizer=Adam(0.001), metrics=['mse'])
    return model

def build_lstm(input_shape):
    model = Sequential([
        LSTM(64, input_shape=input_shape),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1, activation='linear')
    ])
    model.compile(loss='mean_squared_error', optimizer=Adam(0.001), metrics=['mse'])
    return model

# =========================
# Fungsi Pelatihan + Forecasting
# =========================
def train_model(model_builder, X, y, name="Model"):
    kf = KFold(n_splits=5, shuffle=True, random_state=42)
    mses, r2s = [], []
    last_model = None
    last_y_true = last_y_pred = last_history = None

    for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
        print(f"\nFold {fold+1} - {name}")

        model = model_builder(X.shape[1:])

        early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

        history = model.fit(
            X[train_idx], y[train_idx],
            validation_data=(X[val_idx], y[val_idx]),
            epochs=100,
            batch_size=32,
            callbacks=[early_stop],
            verbose=1
        )

        y_pred = model.predict(X[val_idx]).flatten()
        mse = mean_squared_error(y[val_idx], y_pred)
        r2 = r2_score(y[val_idx], y_pred)

        mses.append(mse)
        r2s.append(r2)

        print(f"{name} MSE Fold {fold+1}: {mse:.4f}")
        print(f"{name} R2 Fold {fold+1}: {r2:.4f}")
        print(f"Epoch digunakan: {len(history.history['loss'])}")

        last_model = model
        last_y_true = y[val_idx]
        last_y_pred = y_pred
        last_history = history

    print(f"\n{name} - Mean MSE: {np.mean(mses):.4f}")
    print(f"{name} - Mean R2 : {np.mean(r2s):.4f}")

    y_true_actual = scaler_y.inverse_transform(last_y_true.reshape(-1, 1)).flatten()
    y_pred_actual = scaler_y.inverse_transform(last_y_pred.reshape(-1, 1)).flatten()

    plt.figure(figsize=(12, 5))
    plt.plot(y_true_actual[-50:], label='Actual')
    plt.plot(y_pred_actual[-50:], label='Predicted')
    plt.title(f"{name} - Prediksi vs Aktual (50 data terakhir)")
    plt.xlabel("Index")
    plt.ylabel("Harga")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 4))
    plt.plot(last_history.history['loss'], label='Train Loss')
    plt.plot(last_history.history['val_loss'], label='Val Loss')
    plt.title(f"{name} - Loss Curve")
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    metrics = ['MSE', 'R2']
    values = [mses, r2s]

    plt.figure(figsize=(10, 4))
    for i, (metric, vals) in enumerate(zip(metrics, values)):
        plt.subplot(1, 2, i + 1)
        sns.barplot(x=[f"Fold {j+1}" for j in range(len(vals))], y=vals, palette='Blues')
        plt.axhline(np.mean(vals), color='red', linestyle='--', label=f'Rata-rata: {np.mean(vals):.4f}')
        plt.title(f'{name} - {metric} per Fold')
        plt.ylabel(metric)
        plt.xticks(rotation=45)
        plt.legend()
        plt.grid(True)
    plt.tight_layout()
    plt.show()

    # =========================
    # Forecasting 7 Hari ke Depan
    # =========================
    print(f"\n{name} - Forecasting Harga 7 Hari ke Depan:")

    current_seq = X[-1]  # Ambil sequence terakhir (shape: window_size, features)
    forecasts = []

    for i in range(7):
        input_seq = np.expand_dims(current_seq, axis=0)  # (1, window_size, num_features)
        pred_scaled = last_model.predict(input_seq, verbose=0)[0][0]
        forecasts.append(pred_scaled)

        # Simulasikan input berikutnya
        next_input = current_seq[-1].copy()
        current_seq = np.vstack([current_seq[1:], next_input])

    harga_7_hari = scaler_y.inverse_transform(np.array(forecasts).reshape(-1, 1)).flatten()

    print("Harga prediksi 7 hari ke depan:")
    for i, h in enumerate(harga_7_hari, 1):
        print(f"Hari ke-{i}: {h:.2f}")

    plt.figure(figsize=(8, 4))
    plt.plot(range(1, 8), harga_7_hari, marker='o')
    plt.title(f"{name} - Prediksi Harga 7 Hari ke Depan")
    plt.xlabel("Hari ke-")
    plt.ylabel("Harga (dalam skala asli)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# =========================
# Jalankan model
# =========================
train_model(build_rnn, X_seq, y_seq, name="Simple RNN")
train_model(build_lstm, X_seq, y_seq, name="LSTM")
```
