```python
from google.colab import files
uploaded = files.upload()
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, LSTM, SimpleRNN
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
# Baca file
filename = list(uploaded.keys())[0]
data = pd.read_csv(filename)

# Tampilkan kolom
print(" Kolom tersedia:", data.columns.tolist())

# Coba deteksi kolom harga
harga_cols = ['Close', 'Adj Close', 'Price', 'Last', 'Value']
for col in harga_cols:
    if col in data.columns:
        data['Price'] = data[col]
        print(f" Menggunakan kolom harga: {col}")
        break
else:
    raise ValueError(" Tidak ditemukan kolom harga yang valid!")

# Konversi tanggal jika ada
if 'Date' in data.columns:
    data['Date'] = pd.to_datetime(data['Date'], errors='coerce')
    data = data.dropna(subset=['Date'])
    data = data.sort_values('Date')

# ==========================
#  EDA (Exploratory Data Analysis)
# ==========================

# Statistik dasar
print("\n Statistik Deskriptif:")
print(data.describe())

# Korelasi antar fitur
plt.figure(figsize=(8, 6))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title(" Korelasi antar Fitur")
plt.show()

# Visualisasi harga emas
plt.figure(figsize=(12, 5))
plt.plot(data['Price'], label='Harga Emas')
plt.title(' Harga Emas Seiring Waktu')
plt.xlabel('Index')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()

# Distribusi volume
if 'Volume' in data.columns:
    plt.figure(figsize=(6, 4))
    sns.histplot(data['Volume'], kde=True)
    plt.title(' Distribusi Volume Transaksi')
    plt.show()
data['Label'] = data['Price'].diff().fillna(0).apply(lambda x: 1 if x > 0 else 0)
# Fitur input
features = ['Open', 'High', 'Low', 'Volume']
target = 'Label'

# Normalisasi
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(data[features])
y = data[target].values

# Buat sequence
def create_sequences(X, y, window_size=10):
    Xs, ys = [], []
    for i in range(len(X) - window_size):
        Xs.append(X[i:i+window_size])
        ys.append(y[i+window_size])
    return np.array(Xs), np.array(ys)

window_size = 10
X_seq, y_seq = create_sequences(X_scaled, y, window_size)
def build_rnn(input_shape):
    model = Sequential([
        SimpleRNN(64, input_shape=input_shape),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1, activation='sigmoid')
    ])
    model.compile(loss='binary_crossentropy', optimizer=Adam(0.001), metrics=['accuracy'])
    return model

def build_lstm(input_shape):
    model = Sequential([
        LSTM(64, input_shape=input_shape),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1, activation='sigmoid')
    ])
    model.compile(loss='binary_crossentropy', optimizer=Adam(0.001), metrics=['accuracy'])
    return model
def train_model(model_builder, X, y, name="Model"):
    kf = KFold(n_splits=5, shuffle=True, random_state=42)
    accs = []
    last_y_true = last_y_pred = last_history = None

    for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
        print(f"\n Fold {fold+1} - {name}")
        model = model_builder(X.shape[1:])
        early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

        history = model.fit(
            X[train_idx], y[train_idx],
            validation_data=(X[val_idx], y[val_idx]),
            epochs=100,
            batch_size=32,
            callbacks=[early_stop],
            verbose=1
        )

        y_pred_prob = model.predict(X[val_idx])
        y_pred = (y_pred_prob > 0.5).astype(int)
        acc = accuracy_score(y[val_idx], y_pred)

        accs.append(acc)
        print(f"{name} Accuracy Fold {fold+1}: {acc:.4f}")
        print(f" Epoch digunakan: {len(history.history['loss'])}")

        last_y_true = y[val_idx]
        last_y_pred = y_pred
        last_history = history

    print(f"\n Mean Accuracy for {name}: {np.mean(accs):.4f}")

    # Grafik Prediksi
    plt.figure(figsize=(10, 4))
    plt.plot(last_y_true, label='Actual')
    plt.plot(last_y_pred, label='Predicted', alpha=0.7)
    plt.title(f"{name} - Actual vs Predicted (Last Fold)")
    plt.legend()
    plt.grid(True)
    plt.show()

    # Grafik Loss
    plt.figure(figsize=(8, 4))
    plt.plot(last_history.history['loss'], label='Train Loss')
    plt.plot(last_history.history['val_loss'], label='Val Loss')
    plt.title(f"{name} - Training vs Validation Loss")
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    plt.show()
train_model(build_rnn, X_seq, y_seq, name="Simple RNN")
train_model(build_lstm, X_seq, y_seq, name="LSTM")
```
